### Imports

```python
from aepybamm import get_params, pybamm_print_tools
```

### Print model equations

We use the Molicel P45B parameter set with a blended negative electrode. `get_params()` is called for three model configurations (the default DFN, SPMe, and DFN with a lumped thermal model), and `pybamm_print_tools.as_string()` prints the equations of each resulting model.

```python
fp = "params/Molicel_P45B_bpx.json"

model_inputs = {
    "base": {},
    "SPMe": {"model_type": "SPMe"},
    "lumped thermal": {
        "htc_ext": 1,
        "extra_model_opts": {"thermal": "lumped"},
    },
}

for name, inputs in model_inputs.items():
    _, model = get_params(
        fp,
        blended_electrode=(True, False),  # neg, pos
        **inputs,
    )

    print(f"Printing model equations: {name}\n---\n")
    print(pybamm_print_tools.as_string(model))
    print("\n")
```
